```python
#change name : could have ignoe minuscule and maj
df.loc[df["pnns_groups_1"]=='salty-snacks']='Salty snacks'
df.loc[df["pnns_groups_1"]=='sugary-snacks']='Sugary snacks'
df.loc[df["pnns_groups_1"]=='fruits-and-vegetables']='Fruits and vegetables'
df.loc[df["pnns_groups_1"]=='cereals-and-potatoes']='Cereals and potatoes'
df["pnns_groups_1"].unique()
```

```python
# Peux t on eventuellement changer les nunknown de 1 avec des valeurs renseignées du 2?
dd=df.loc[df["pnns_groups_1"]=='unknown',["pnns_groups_2"]]
dd["pnns_groups_2"].unique()
#non tous les unknown de 1 ont comme valeur dans le 2 unknow
```

```python
df["pnns_groups_1"].replace(r"unknown", np.nan, inplace=True)
df["pnns_groups_2"].replace(r"unknown", np.nan, inplace=True)
df["pnns_groups_1"].unique()
```

```python
# Peux t on eventuellement changer les NAN de 1 avec des valeurs renseignées du 2?
dd2=df.loc[df["pnns_groups_2"]=='nan',["pnns_groups_1"]]
dd2["pnns_groups_1"].unique()
#NON
```

```python
#compare les 2 col pnns group1 et 2
#df['group1-2'] = np.where(df['pnns_groups_1'] == df['pnns_groups_2'])
```

```python
df_1_med=df.groupby(by=['pnns_groups_1'])[list_col_float].median().round(2)
df_1_med.reset_index(inplace=True)
df_1_med
```

```python
df_group1=df.groupby(by=['pnns_groups_1'])[list_col_float].mean().round(2)
df_group1.reset_index(inplace=True)
df_group1
```

```python
df_group2=df.groupby(by=['pnns_groups_2'])[list_col_float].mean().round(2)
df_group2.reset_index(inplace=True)
df_group2
```

```python
#test remplacer avec valeur
s=df.groupby("pnns_groups_2")['energy_100g'].agg(lambda x:x.mode(dropna=False).iat[0])
df['energy_100g']=df['energy_100g'].fillna(df["pnns_groups_2"].map(s))
df['energy_100g'].isna().count()
```

```python
df.loc[df["pnns_groups_2"]== "Alcoholic beverages",'fat_100g'].fillna(df_group2.loc[df_group2["pnns_groups_2"]=="Alcoholic beverages",'fat_100g'], inplace=True)
df["fat_100g"].isna().count()
```

```python
list_col_group2=df_group2["pnns_groups_2"].unique()
df2=df
```

```python
# Boucle for
for i in list_col_group2:
    for j in list_col_float:
        df.loc[df["pnns_groups_2"]== i,j].fillna(df_group2.loc[df_group2["pnns_groups_2"]==i,j], inplace=True)
df["saturated-fat_100g"].isna().count()
```

```python
s2=df.groupby("pnns_groups_2")['fat_100g'].agg(lambda x:x.mode(dropna=False).iat[0])
df['fat_100g']=df['fat_100g'].fillna(df["pnns_groups_2"].map(s2))
df['fat_100g'].isna().count()
```
